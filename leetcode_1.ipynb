```python
# lets define a function called twoSum
def twoSum(nums, target):
    dic = {}
    for i, num in enumerate(nums):#loops through the list keeping track of index and elements
        if target - num in dic:
            return [dic[target - num], i]
        dic[num] = i
    return []
```
