**Problem** : Given an array of numbers e,g [1,2,4,6,8,6] and a target e.g 5 <br> return an array showing the  two indexes whose element sum to 5
<hr>

**1.Brute-Force-approach**


---



> ForEach element in the array , take target - element. and if result is in the array , <br> you return [ index of element, index of (target - element) ]






````python
def twoSum(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i, j]
````

**time compexity** : $O(n^2)$

````python
#optimized approach

# lets define a function called twoSum
def twoSum(nums, target):
    dic = {}
    for i, num in enumerate(nums):#loops through the list keeping track of index and elements
        if target - num in dic:
            return [dic[target - num], i]
        dic[num] = i
    return []
````

**Time complexity** : $O(n)$ <br> **Space complxity**:$O(1)$

**enumerate() function** it is a built-in function that adds a counter to an iterable <br> (like a list or tuple) and returns it as an enumerate object (which yields pairs of index and value). <br>


```python
nums = [10,20,30]

for num in nums:

    i += 1

    print(i, num)

```
